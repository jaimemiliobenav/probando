### Caso aplicado: Control de lotes de producción de yogurt

Una planta agroindustrial necesita organizar información básica de sus lotes de yogurt. Para cada lote se registra su **código, sabor, cantidad producida, unidades rechazadas y costo unitario de producción**. Se desea representar cada lote mediante objetos y automatizar algunos cálculos de control.

## 1. Clase `LoteYogurt` — 8 puntos

Defina la clase con los atributos:
`codigo`, `sabor`, `producidas`, `rechazadas`, `costo_unitario`.

Implemente:
- `tasa_rechazo()`: porcentaje de rechazadas respecto a producidas.
- `unidades_conformes()`: producidas − rechazadas.
- `costo_lote()`: producidas × costo unitario.
- `estado_calidad()`: `"REVISAR"` si la tasa de rechazo es mayor a 5%; de lo contrario `"ACEPTABLE"`.

```python
# PREGUNTA 1
# Complete aquí la clase LoteYogurt

class LoteYogurt:
    # Escriba su código
    def __init__(self, codigo, sabor, producidas, rechazadas, costo_unitario):
      self.codigo = codigo
      self.sabor = sabor
      self.producidas = producidas
      self.rechazadas = rechazadas
      self.costo_unitario = costo_unitario

    def tasa_rechazo(self):
      if self.producidas == 0:
        return 0
      return (self.rechazadas / self.producidas) * 100

    def unidades_conformes(self):
      return self.producidas - self.rechazadas

    def costo_lote(self):
      return self.producidas * self.costo_unitario

    def estado_calidad(self):
      if self.tasa_rechazo() > 5:
        return "REVISAR"
      else:
        return "ACEPTABLE"
```

## 2. Creación y uso de objetos — 6 puntos

Cree los siguientes lotes:

| Código | Sabor | Producidas | Rechazadas | Costo unitario |
|---|---|---:|---:|---:|
| L001 | Fresa | 1200 | 36 | 1.80 |
| L002 | Natural | 950 | 67 | 1.65 |
| L003 | Durazno | 1500 | 45 | 1.90 |

1. Guárdelos en una lista llamada `lotes`.
2. Recorra la lista e imprima código, sabor, tasa de rechazo y estado.
3. Determine cuál tiene la **mayor cantidad de unidades conformes** e imprima código y cantidad.

```python
# PREGUNTA 2
# Cree los objetos y resuelva lo solicitado.
L1 = LoteYogurt("L001", "Fresa", 1200, 36, 1.80)
L2 = LoteYogurt("L002", "Natural", 950, 67, 1.65)
L3 = LoteYogurt("L003", "Durazno", 1500, 45, 1.90)

lotes = [L1, L2, L3]

# Recorra la lista e imprima código, sabor, tasa de rechazo y estado.

for lote in lotes:
  print(f"Código: {lote.codigo}, Sabor: {lote.sabor}, Tasa de rechazo: {lote.tasa_rechazo()}, Estado: {lote.estado_calidad()}")


lote_mayor = lotes[0]
for lote in lotes:
  if lote.unidades_conformes() > lote_mayor.unidades_conformes():
    lote_mayor = lote

print(f"Código con mayor conformes: {lote_mayor.codigo}")
print(f"Cantidad conformes: {lote_mayor.unidades_conformes()}")

```

## 3. Herencia — 4 puntos

Cree `LoteExportacion`, que herede de `LoteYogurt`, y agregue:
- `pais_destino`
- `costo_certificacion`

Use `super()` para reutilizar el constructor de la clase base.

Luego cree:
`E001`, sabor Natural, 2000 producidas, 80 rechazadas, costo unitario 1.75, destino Chile y costo de certificación 650.

Imprima su código, país de destino y estado de calidad.

```python
# PREGUNTA 3
# Complete aquí.

class LoteExportacion(LoteYogurt):

  def __init__(self,codigo,sabor,producidas,rechazadas,costo_unitario,pais_destino,costo_certificacion,):

    super().__init__(codigo, sabor, producidas, rechazadas, costo_unitario)
    self.pais_destino = pais_destino
    self.costo_certificacion = costo_certificacion

lote_exp = LoteExportacion("E001", "Natural", 2000, 80, 1.75, "Chile", 650)

print("Código:", lote_exp.codigo)
print("País de destino:", lote_exp.pais_destino)
print("Estado de calidad:", lote_exp.estado_calidad())
```

## 4. Comprensión de POO — 2 puntos

**Explique brevemente:** ¿por qué `LoteExportacion` puede utilizar `estado_calidad()` aunque ese método no se programe nuevamente dentro de la clase hija?

**Respuesta:**  
>Porque al aplicar herencia (class LoteExportacion(LoteYogurt):), la clase hija recibe automáticamente todo lo que ya tiene la clase padre. Así que no hace falta volver a escribir estado_calidad(); dado que la clase hija ya lo "sabe hacer" por herencia directa uwu.
